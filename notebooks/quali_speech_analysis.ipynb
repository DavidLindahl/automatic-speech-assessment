# QualiSpeech Dataset Analysis

This notebook downloads and analyzes the `tsinghua-ee/QualiSpeech` dataset from Hugging Face, using a local `data-temp` cache directory. It also saves split-level metadata CSV exports for easy inspection.

```python
from pathlib import Path
import pandas as pd
from datasets import load_dataset

DATA_ROOT = Path('data-temp')
CACHE_DIR = DATA_ROOT / 'hf_cache'
DATA_ROOT.mkdir(parents=True, exist_ok=True)
CACHE_DIR.mkdir(parents=True, exist_ok=True)

dataset = load_dataset('tsinghua-ee/QualiSpeech', cache_dir=str(CACHE_DIR))
print('Dataset loaded into cache_dir:', CACHE_DIR.resolve())
print(dataset)
```

```python
pd.set_option('display.max_columns', None)
rating_cols = [
    'Speed',
    'Naturalness',
    'Background noise',
    'Distortion',
    'Listening effort',
    'Continuity',
    'Overall quality',
    'Feeling of voice',
]

dfs = {}
for split, ds in dataset.items():
    df = ds.to_pandas()
    dfs[split] = df
    print(f'--- {split} ---')
    print('rows:', len(df))
    print('columns:', df.columns.tolist())
    display(df.head(1))
    display(df[rating_cols].describe().transpose())
```

```python
for split, df in dfs.items():
    print(f'Top 5 Overall quality examples in {split}')
    display(df.sort_values('Overall quality', ascending=False).head(5))

    print(f'Bottom 5 Overall quality examples in {split}')
    display(df.sort_values('Overall quality', ascending=True).head(5))
```

```python
for split, df in dfs.items():
    out_file = DATA_ROOT / f'QualiSpeech_{split}.csv'
    df.to_csv(out_file, index=False)
    print('Saved metadata CSV:', out_file)
```
